# 3. Consistencia de variables categóricas 
    Revisamos todas las columnas que son cadenas de texto

```python
columnas = [
    "Gender",
    "Branch",
    "City",
    "Customer type",
    "Product line",
    "Payment"
]

for columna in columnas:
    print(f"\n--- {columna} ---")
    display(df[columna].value_counts(dropna=False))
```

## Validación y limpieza de datos

Se han realizado las siguientes comprobaciones de calidad:

- **Valores nulos:** no se han detectado valores ausentes mediante `isna()`.
- **Cadenas vacías:** no se han encontrado celdas vacías (`""`) ni celdas que contengan exactamente un espacio (`" "`).
- **Duplicados:** no existen filas completamente duplicadas ni identificadores de factura (`Invoice ID`) repetidos.
- **Consistencia categórica:** se han revisado las columnas `Gender`, `Branch`, `City`, `Customer type`, `Product line` y `Payment` mediante `value_counts()`. No se han observado inconsistencias evidentes en sus categorías.
- **Fechas:** se ha convertido la columna `Date` de texto a tipo `datetime`, utilizando el formato original día/mes/año.

**Conclusión provisional:** los datos presentan una buena calidad inicial. No ha sido necesario eliminar registros ni corregir las categorías revisadas.
