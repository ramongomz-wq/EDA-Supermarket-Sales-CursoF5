### Validación de coherencia en transacciones con cantidades elevadas

Durante el análisis exploratorio se han identificado transacciones con cantidades de productos superiores a las habituales.

Para comprobar su consistencia, se analizarán las transacciones cuya columna `Quantity` sea superior a **14 unidades**, valor correspondiente al tercer cuartil (Q3).

Se verificará que el precio unitario calculado a partir del importe total, descontando los impuestos, coincida con el precio unitario registrado.

**Fórmula de validación:**

\[
\text{Precio unitario calculado} = \frac{\text{Total} - \text{Tax 5\%}}{\text{Quantity}}
\]

La comparación devolverá valores booleanos (`True` o `False`), que se contabilizarán para identificar posibles discrepancias.

**Nota:** las diferencias pueden deberse a la precisión de los números decimales y no necesariamente a errores en los datos.

```python
compras_altas = df.loc[df["Quantity"] > 14]

(
    (compras_altas["Total"] - compras_altas["Tax 5%"])
    / compras_altas["Quantity"]
    == compras_altas["Unit price"]
).value_counts()
```

```python
display(
    ((compras_altas["Total"] - compras_altas["Tax 5%"])
    / compras_altas["Quantity"]).head()
)

display(compras_altas["Unit price"].head())
```

```python
display(df.loc[[1000]])
```

```python
df["importe_esperado"] = df["Unit price"] * df["Quantity"]

df["importe_real_sin_impuestos"] = df["Total"] - df["Tax 5%"]

df["impuesto_esperado"] = df["importe_esperado"] * 0.05

display(df[[
    "importe_esperado",
    "impuesto_esperado",
    "Tax 5%",
    "importe_real_sin_impuestos"
]].head(25))
```

### Investigación de posibles discrepancias en los importes e impuestos

Durante la validación de las variables económicas se han detectado diferencias entre los importes calculados a partir de `Unit price` y `Quantity` y los importes registrados en el dataset.

Inicialmente se analizaron las transacciones con `Quantity > 14` (tercer cuartil), pero posteriormente se amplió la investigación a las **1.400 facturas** para comprobar si las discrepancias siguen algún patrón.

Para ello, se crearon tres variables auxiliares:

- `importe_esperado`: resultado de multiplicar `Unit price` por `Quantity`.
- `impuesto_esperado`: 5 % del importe esperado.
- `importe_real_sin_impuestos`: resultado de restar `Tax 5%` a `Total`.

**Hallazgos preliminares:**

En las primeras 25 transacciones examinadas, los impuestos registrados no coinciden exactamente con el 5 % del importe esperado. También se observan diferencias entre los importes teóricos y los registrados.

Una posible explicación sería la existencia de **descuentos por volumen u otros ajustes comerciales**, aunque el dataset no contiene información suficiente para confirmarlo.

En la primera transacción examinada, el importe real sin impuestos es 658 y el impuesto registrado es 33. El 5 % de 658 equivale a 32,90, una cantidad muy próxima al impuesto registrado.

**Nueva hipótesis:**

El impuesto podría estar calculándose sobre el importe final antes de impuestos, posiblemente después de aplicar descuentos, y redondeándose a números enteros.

**Siguiente comprobación:**

Se calculará el 5 % de `importe_real_sin_impuestos` y se comparará con `Tax 5%` para investigar si existe un patrón consistente.

Los resultados permitirán evaluar esta hipótesis, pero no demostrarán por sí solos la existencia de descuentos.

```python
df["impuesto_sobre_importe_real"] = df["importe_real_sin_impuestos"] * 0.05

display(df[[
    "importe_real_sin_impuestos",
    "impuesto_sobre_importe_real",
    "Tax 5%"
]].head(20))
```

```python
df["diferencia_impuesto"] = (
    df["Tax 5%"] - df["impuesto_sobre_importe_real"]
)

display(df["diferencia_impuesto"].describe())
```

```python
(df["diferencia_impuesto"].abs() > 1).sum()
```

### Conclusiones de la validación de importes e impuestos

Durante el análisis exploratorio se han identificado discrepancias entre los importes teóricos calculados a partir de `Unit price` y `Quantity` y los valores registrados en `Total` y `Tax 5%`.

Se ha investigado la posibilidad de que existan descuentos comerciales y de que los impuestos se calculen sobre importes ajustados. Sin embargo, el dataset no dispone de información suficiente para confirmar estas hipótesis.

Al comparar el impuesto registrado con el 5 % del importe real antes de impuestos:

- **1.070 transacciones (76,43 %)** presentan diferencias absolutas de una unidad monetaria o menos.
- **330 transacciones (23,57 %)** presentan diferencias absolutas superiores a una unidad monetaria.

Estas diferencias sugieren posibles inconsistencias o reglas de cálculo no documentadas.

**Decisión:** se conservarán las 1.400 transacciones sin modificar sus importes originales. Las discrepancias se documentarán como una limitación del dataset y deberán tenerse en cuenta al interpretar los indicadores económicos.
